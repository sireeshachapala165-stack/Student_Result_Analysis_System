```python
import pandas as pd

# Read CSV file
data = pd.read_csv("student_marks.csv")

# Show dataset
print("Student Data:")
print(data)

# Create Total Marks column
data["Total"] = (
    data["Maths"] +
    data["English"] +
    data["Science"]
)

# Create Percentage column
data["Percentage"] = data["Total"] / 3

# Find topper
topper = data.loc[data["Total"].idxmax()]

print("\nTopper Student:")
print(topper["Name"])

# Average marks for each subject
print("\nAverage Maths Marks:")
print(data["Maths"].mean())

print("\nAverage English Marks:")
print(data["English"].mean())

print("\nAverage Science Marks:")
print(data["Science"].mean())

# Failed students (Total below 150)
failed_students = data[data["Total"] < 150]

print("\nFailed Students:")
print(failed_students["Name"])

# Highest total marks
print("\nHighest Total Marks:")
print(data["Total"].max())

# Show final table
print("\nFinal Student Report:")
print(data)
```

```python
# Save final report
data.to_csv("student_final_report.csv", index=False)

print("File saved successfully!")
```
